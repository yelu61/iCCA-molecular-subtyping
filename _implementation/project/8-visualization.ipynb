# Step 8 — Reproduce the selected manuscript figures

Updated 2026-11-12. This notebook uses the same public entry points as `notebooks/Reproduce.ipynb`. Supply the companion frozen-input bundle and choose a fresh output directory. The selected manuscript workflow comprises 37 renderer jobs and three final drivers.


```python
from pathlib import Path
import json
import subprocess
import sys

candidates = []
for folder in [Path.cwd(), *Path.cwd().parents]:
    candidates.extend([folder, folder / "code_release/iCCA-molecular-subtyping"])
CODE_REPO = next((p for p in candidates if (p / "run.py").is_file()
                  and (p / "config/analysis_stages.json").is_file()), None)
if CODE_REPO is None:
    raise FileNotFoundError("Open this notebook from the research project or source-code repository.")
catalog = json.loads((CODE_REPO / "config/analysis_stages.json").read_text())
```

```python
DATA_ROOT = None  # Set to Path("/path/to/data_bundle")
OUTPUT_ROOT = None  # Set to Path("/path/to/new_output")
RUN_FIGURES = False
```

```python
if DATA_ROOT is None or OUTPUT_ROOT is None:
    print("Set DATA_ROOT and OUTPUT_ROOT for the frozen-input reproduction workflow.")
else:
    data = str(Path(DATA_ROOT).expanduser().resolve())
    output = str(Path(OUTPUT_ROOT).expanduser().resolve())
    commands = [
        [sys.executable, str(CODE_REPO / "run.py"), "verify", "--data-root", data],
        [sys.executable, str(CODE_REPO / "run.py"), "figures", "--data-root", data, "--output", output],
        [sys.executable, str(CODE_REPO / "check_results.py"), "--run-dir", output, "--data-root", data],
    ]
    for command in commands:
        print(json.dumps(command))
        if RUN_FIGURES:
            subprocess.run(command, cwd=CODE_REPO, check=True)
```
