```python
from pathlib import Path
import json
import subprocess
import sys

candidates = []
for folder in [Path.cwd(), *Path.cwd().parents]:
    candidates.extend([folder, folder / "code_release/iCCA-molecular-subtyping"])
CODE_REPO = next((p for p in candidates if (p / "run.py").is_file()
                  and (p / "config/analysis_stages.json").is_file()), None)
if CODE_REPO is None:
    raise FileNotFoundError("Open this notebook from the research project or source-code repository.")
catalog = json.loads((CODE_REPO / "config/analysis_stages.json").read_text())

STAGES = ['singlecell_programs', 'singlecell_display_sources', 'singlecell_cnv_cases', 'singlecell_score_baseline']
for name in STAGES:
    print(name, catalog[name]["entry"])
```

## Configure the analysis

Choose a stage and an independent project directory containing its required upstream inputs. The stage documentation and external interfaces are in the code repository. `EXECUTE=False` prepares the command and stages code using the existing CLI behavior.


```python
STAGE = 'singlecell_programs'
PROJECT_ROOT = None  # Set to Path("/path/to/independent_project")
EXECUTE = False
```

```python
if PROJECT_ROOT is None:
    print("Set PROJECT_ROOT to an independent directory with the required inputs.")
else:
    if STAGE not in STAGES:
        raise ValueError("Choose one of the displayed stages.")
    command = [sys.executable, str(CODE_REPO / "run.py"), "analysis",
               "--stage", STAGE, "--project-root", str(Path(PROJECT_ROOT).expanduser().resolve())]
    if EXECUTE:
        command.append("--execute")
    print(json.dumps(command))
    subprocess.run(command, cwd=CODE_REPO, check=True)
```
